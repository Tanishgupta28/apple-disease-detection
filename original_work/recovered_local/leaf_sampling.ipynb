```python
import os
import shutil
import random
import cv2

# -------------------------------
# Configuration
# -------------------------------
original_dataset = 'Apple_disease'            # Your source dataset folder
new_dataset = 'sampled_dataset'               # New destination folder
max_images_per_class = 125                    # Max number of images per class
resize_images = False                         # Set True to resize to 256x256
image_size = (256, 256)                       # Resize size if enabled

# -------------------------------
# List of class folders
# -------------------------------
classes = ['scab', 'black_rot', 'cedar_rust', 'healthy']

# -------------------------------
# Create new base folder
# -------------------------------
os.makedirs(new_dataset, exist_ok=True)

for cls in classes:
    src_folder = os.path.join(original_dataset, cls)
    dst_folder = os.path.join(new_dataset, cls)
    os.makedirs(dst_folder, exist_ok=True)

    # ✅ Collect all supported image files
    all_imgs = [f for f in os.listdir(src_folder)
                if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

    print(f"\n📂 Class '{cls}' — Found {len(all_imgs)} image(s) in: {src_folder}")

    if len(all_imgs) == 0:
        print(f"⚠️ No valid images found in '{src_folder}' — skipping.")
        continue

    sampled_imgs = random.sample(all_imgs, min(len(all_imgs), max_images_per_class))

    copied = 0
    for img_name in sampled_imgs:
        src_path = os.path.join(src_folder, img_name)
        dst_path = os.path.join(dst_folder, img_name)

        # Skip if file is missing or empty
        if not os.path.exists(src_path) or os.path.getsize(src_path) == 0:
            print(f"⚠️ Skipping invalid file: {src_path}")
            continue

        # Check if the image is loadable
        image = cv2.imread(src_path)
        if image is None:
            print(f"❌ Cannot read image (possibly corrupted): {src_path}")
            continue

        # Resize if needed
        if resize_images:
            try:
                image = cv2.resize(image, image_size)
                cv2.imwrite(dst_path, image)
            except Exception as e:
                print(f"❌ Resize failed for {src_path}: {e}")
                continue
        else:
            try:
                shutil.copyfile(src_path, dst_path)
            except Exception as e:
                print(f"❌ Copy failed for {src_path}: {e}")
                continue

        copied += 1

    print(f"✅ Copied {copied} image(s) to: {dst_folder}")

print("\n🎉 Done! Sampled dataset created in:", new_dataset)
```
