```python
## now we are going to reduce image resolution 
from PIL import Image
image_folder = r"yolo_dataset/images"

# Target size for YOLO
target_size = (416, 416)

# Loop through every image in the folder
for img_name in os.listdir(image_folder):
    img_path = os.path.join(image_folder, img_name)
    
    if os.path.isfile(img_path):
        try:
            img = Image.open(img_path)
            img = img.resize(target_size)
            img.save(img_path, optimize=True, quality=85)  # overwrite with compressed image
        except Exception as e:
            print(f"Error processing {img_path}: {e}")
print("All images resized to 416x416 and compressed")
```
